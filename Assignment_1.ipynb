```python
# Create empty output image
transformed = np.zeros_like(image, dtype=np.uint8)

# Apply transformation based on ranges
for i in range(image.shape[0]):
    for j in range(image.shape[1]):
        pixel = image[i, j]
        if pixel < 50:
            new_val = pixel
        elif pixel < 100:
            new_val = 2 * pixel - 50
        elif pixel < 150:
            new_val = 2.5 * pixel - 150
        else:
            new_val = pixel + 100
        # Clip to 0–255
        transformed[i, j] = np.clip(new_val, 0, 255)

# Show results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("Transformed Image")
plt.imshow(transformed, cmap='gray')
plt.show()
```

```python
#(b) Apply the aforementioned intensity transformation to the saturation plane.


# Apply intensity transformation to saturation
def vibrance_boost(s_plane, a=0.8, sigma=70):
    s = s_plane.astype(np.float32)
    x = s.copy()
    boost = a * 128 * np.exp(-((x - 128) ** 2) / (2 * sigma ** 2))
    enhanced = np.minimum(x + boost, 255)
    return enhanced.astype(np.uint8)
```

```python

#(c) Adjust a to get a visually pleasing output. Report the value of a.

# Adjust `a` and report
#a_val = 0.8
a_val = 0.6
#a_val = 1
s_enhanced = vibrance_boost(s, a=a_val)
```

```python
#(d) Recombine the three planes.
# Step (d): Recombine and convert to RGB
hsv_enhanced = cv2.merge([h, s_enhanced, v])
img_enhanced_rgb = cv2.cvtColor(hsv_enhanced, cv2.COLOR_HSV2RGB)
```

```python
#(a) Using the existing filter2D to Sobel filter the image.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load grayscale image
img = cv2.imread('images/einstein.png', cv2.IMREAD_GRAYSCALE)

# Define Sobel kernel for x-direction
sobel_kernel_x = np.array([[1, 0, -1],
                           [2, 0, -2],
                           [1, 0, -1]])

# Apply filter2D
sobel_x = cv2.filter2D(img, -1, sobel_kernel_x)

plt.imshow(sobel_x, cmap='gray')
plt.title('Sobel Filter using filter2D (X-direction)')
plt.axis('off')
plt.show()

```

```python
#(b) Write your own code to Sobel filter the image.
# Padding the image
padded = cv2.copyMakeBorder(img, 1, 1, 1, 1, cv2.BORDER_REPLICATE)

# Initialize empty output
sobel_manual = np.zeros_like(img, dtype=np.float32)

# Manually convolve with the Sobel kernel
for i in range(1, padded.shape[0] - 1):
    for j in range(1, padded.shape[1] - 1):
        region = padded[i - 1:i + 2, j - 1:j + 2]
        sobel_manual[i - 1, j - 1] = np.sum(region * sobel_kernel_x)

# Normalize for display
sobel_manual = cv2.convertScaleAbs(sobel_manual)

plt.imshow(sobel_manual, cmap='gray')
plt.title('Sobel Filter (Manual Code)')
plt.axis('off')
plt.show()
```

```python
#(c) Using the property

# 1D filters
k1 = np.array([[1], [2], [1]])       # Vertical
k2 = np.array([[1, 0, -1]])          # Horizontal

# Step 1: Apply vertical (column) filter
intermediate = cv2.filter2D(img, -1, k1)

# Step 2: Apply horizontal (row) filter
sobel_separable = cv2.filter2D(intermediate, -1, k2)

plt.imshow(sobel_separable, cmap='gray')
plt.title('Sobel Filter (Using Separable Kernels)')
plt.axis('off')
plt.show()
```

```python
def nearest_neighbor_zoom(img, scale_factor):
    height, width = img.shape[:2]
    new_height = int(height * scale_factor)
    new_width = int(width * scale_factor)

    # Fixed line
    zoomed_img = np.zeros((new_height, new_width), dtype=img.dtype)

    for i in range(new_height):
        for j in range(new_width):
            src_x = int(i / scale_factor)
            src_y = int(j / scale_factor)
            zoomed_img[i, j] = img[src_x, src_y]

    return zoomed_img
```

```python
#(b) bilinear interpolation.

def bilinear_zoom(img, scale_factor):
    height, width = img.shape[:2]
    new_height = int(height * scale_factor)
    new_width = int(width * scale_factor)
    channels = img.shape[2]

    zoomed_img = np.zeros((new_height, new_width, channels), dtype=np.uint8)

    for i in range(new_height):
        for j in range(new_width):
            x = i / scale_factor
            y = j / scale_factor

            x0 = int(np.floor(x))
            x1 = min(x0 + 1, height - 1)
            y0 = int(np.floor(y))
            y1 = min(y0 + 1, width - 1)

            a = x - x0
            b = y - y0

            for c in range(channels):
                value = (1 - a) * (1 - b) * img[x0, y0, c] + \
                        a * (1 - b) * img[x1, y0, c] + \
                        (1 - a) * b * img[x0, y1, c] + \
                        a * b * img[x1, y1, c]

                zoomed_img[i, j, c] = int(np.clip(value, 0, 255))

    return zoomed_img

# Display
import matplotlib.pyplot as plt
plt.imshow(zoomed_image, cmap='gray')
plt.title(f"Nearest Neighbor Zoom (x{scale})")
plt.axis('off')
plt.show()
```

```python
#(a) Use grabCut to segment the image. Show the final segmentation mask, foreground image, and background image.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread('images/daisy.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Create mask
mask = np.zeros(img.shape[:2], np.uint8)

# Temporary arrays used by grabCut
bgdModel = np.zeros((1, 65), np.float64)
fgdModel = np.zeros((1, 65), np.float64)

# Define rectangle around the flower [x, y, width, height]
height, width = img.shape[:2]
rect = (int(0.1*width), int(0.1*height), int(0.8*width), int(0.8*height))

# Apply grabCut
cv2.grabCut(img_rgb, mask, rect, bgdModel, fgdModel, 5, cv2.GC_INIT_WITH_RECT)

# Create segmentation mask
seg_mask = np.where((mask == 2) | (mask == 0), 0, 1).astype('uint8')

# Extract foreground and background
foreground = img_rgb * seg_mask[:, :, np.newaxis]
background = img_rgb * (1 - seg_mask[:, :, np.newaxis])

# Plot results
plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.imshow(seg_mask * 255, cmap='gray')
plt.title("Segmentation Mask")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(foreground)
plt.title("Foreground Image")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(background)
plt.title("Background Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#(b) Produce an enhanced image with a substantially blurred background. Display the original image alongside the enhanced image

# Apply heavy Gaussian blur to the original image
blurred_img = cv2.GaussianBlur(img_rgb, (45, 45), 0)

# Use the segmentation mask to keep the foreground and blur the background
enhanced_img = np.where(seg_mask[:, :, np.newaxis] == 1, img_rgb, blurred_img)

# Display Original and Enhanced images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(enhanced_img)
plt.title("Enhanced Image (Blurred Background)")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Feather (blur) the segmentation mask slightly
mask_blurred = cv2.GaussianBlur(seg_mask.astype(np.float32), (15, 15), 0)

# Blend foreground and blurred background smoothly
enhanced_smooth = (img_rgb * mask_blurred[:, :, np.newaxis] +
                   blurred_img * (1 - mask_blurred[:, :, np.newaxis])).astype(np.uint8)

# Show improved result
plt.imshow(enhanced_smooth)
plt.title("Enhanced Image with Smooth Edge")
plt.axis("off")
plt.show()
```

```python
#10. Fig. 9 shows a an image of two sapphires placed on a table. The optical axis is perpendicular to the surface of the table.

#(a) Apply a segmentation algorithm to obtain a binary mask for the sapphires.

import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread("images/sapphire.jpg")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Define HSV range for blue sapphires
lower_blue = np.array([100, 50, 50])
upper_blue = np.array([140, 255, 255])

# Create binary mask
mask = cv2.inRange(hsv, lower_blue, upper_blue)

# Display
plt.imshow(mask, cmap='gray')
plt.title("Binary Mask (Segmented)")
plt.axis("off")
plt.show()
```

```python
#(b) Apply a morphological operation to fill the holes.

# Morphological Closing to fill holes
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7, 7))
mask_closed = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

plt.imshow(mask_closed, cmap='gray')
plt.title("After Morphological Closing")
plt.axis("off")
plt.show()
```

```python
#(c) Run connectedComponentsWithStats to obtain the areas in pixels
# Connected components analysis
num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(mask_closed, connectivity=8)

# Extract area of each sapphire (ignore background = index 0)
for i in range(1, num_labels):
    print(f"Sapphire {i}: Area in pixels = {stats[i, cv2.CC_STAT_AREA]}")

```

```python
#(d) If f = 8 mm and the camera lens is 480 mm away from the table surface, compute the actual areas of the sapphires.
f_mm = 8
Z_mm = 480

for i in range(1, num_labels):
    area_px = stats[i, cv2.CC_STAT_AREA]
    actual_area = (area_px * Z_mm**2) / (f_mm**2)
    print(f"Sapphire {i}: Actual area ≈ {actual_area:.2f} mm²")
```
