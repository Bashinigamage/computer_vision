```python
import cv2
import numpy as np
import matplotlib.pyplot as plt



# Load grayscale image
image = cv2.imread('images/emma.jpg', cv2.IMREAD_GRAYSCALE)

print(image)

```

```python
# Create empty output image
transformed = np.zeros_like(image, dtype=np.uint8)

# Apply transformation based on ranges
for i in range(image.shape[0]):
    for j in range(image.shape[1]):
        pixel = image[i, j]
        if pixel < 50:
            new_val = pixel
        elif pixel < 100:
            new_val = 2 * pixel - 50
        elif pixel < 150:
            new_val = 2.5 * pixel - 150
        else:
            new_val = pixel + 100
        # Clip to 0–255
        transformed[i, j] = np.clip(new_val, 0, 255)

# Show results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("Transformed Image")
plt.imshow(transformed, cmap='gray')
plt.show()
```

```python
# ✅ Save the transformed image to disk
cv2.imwrite("images/transformed/transformed_image.jpg", transformed)
```

```python
import matplotlib.pyplot as plt

# Prepare empty outputs
white_matter = np.zeros_like(image, dtype=np.uint8)
gray_matter = np.zeros_like(image, dtype=np.uint8)

# Transformation 1: White Matter Enhancement
for i in range(image.shape[0]):
    for j in range(image.shape[1]):
        pixel = image[i, j]
        if 120 <= pixel <= 200:
            white_matter[i, j] = min(255, (pixel - 120) * 3)
        else:
            white_matter[i, j] = 0

# Transformation 2: Gray Matter Enhancement
for i in range(image.shape[0]):
    for j in range(image.shape[1]):
        pixel = image[i, j]
        if 50 <= pixel <= 120:
            gray_matter[i, j] = min(255, (pixel - 50) * 4)
        else:
            gray_matter[i, j] = 0

# Plotting the results
plt.figure(figsize=(12, 8))

# Original
plt.subplot(1, 3, 1)
plt.title("Original")
plt.imshow(image, cmap='gray')
plt.axis('off')

# White matter enhanced
plt.subplot(1, 3, 2)
plt.title("White Matter Enhanced")
plt.imshow(white_matter, cmap='gray')
plt.axis('off')

# Gray matter enhanced
plt.subplot(1, 3, 3)
plt.title("Gray Matter Enhanced")
plt.imshow(gray_matter, cmap='gray')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#03 (a)
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load the image
img_org = cv2.imread("images/highlights_and_shadows.jpg")
img_rgb = cv2.cvtColor(img_org, cv2.COLOR_BGR2RGB)

# Convert to LAB color space
img_lab = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2Lab)

# Split LAB into L, a, b channels
L, a, b = cv2.split(img_lab)

# Normalize L to [0, 1]
L_norm = L / 255.0


# After converting to float Lab:
img_lab_float = cv2.cvtColor(img_rgb.astype(np.float32) / 255.0, cv2.COLOR_RGB2Lab)
L_float = img_lab_float[:, :, 0]  # Now in range [0, 100]
L_norm = L_float / 100.0

# Gamma correction
gamma = 0.5  # You can adjust this value (example value for brightening)
#gamma = 2.2
gamma = -0.5
L_gamma = np.power(L_norm, gamma)

# Rescale to [0, 255]
L_gamma_corrected = np.uint8(L_gamma * 255)

# Merge and convert back to RGB
lab_corrected = cv2.merge((L_gamma_corrected, a, b))
img_corrected_rgb = cv2.cvtColor(lab_corrected, cv2.COLOR_Lab2RGB)

# Display original and corrected images
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_corrected_rgb)
plt.title(f"Gamma Corrected (γ = {gamma})")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import color

# Load image in RGB
image_bgr = cv2.imread("images/highlights_and_shadows.jpg")
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Convert RGB to L*a*b*
lab_image = color.rgb2lab(image_rgb)
L_channel = lab_image[:, :, 0]

# Save original L for histogram
L_original = L_channel.copy()

# Apply gamma correction (e.g., gamma = 0.5)
#gamma = 0.5
#gamma = 2.2
gamma =  1
L_corrected = ((L_channel / 100) ** gamma) * 100
lab_image[:, :, 0] = L_corrected

# Convert back to RGB
corrected_rgb = color.lab2rgb(lab_image)

# Plot original vs corrected image
plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.title("Original Image")
plt.imshow(image_rgb)
plt.axis("off")

plt.subplot(2, 2, 2)
plt.title(f"Gamma Corrected (γ={gamma})")
plt.imshow(corrected_rgb)
plt.axis("off")

# Plot histograms
plt.subplot(2, 2, 3)
plt.title("Histogram of Original L")
plt.hist(L_original.ravel(), bins=256, color='gray')

plt.subplot(2, 2, 4)
plt.title("Histogram of Gamma Corrected L")
plt.hist(L_corrected.ravel(), bins=256, color='red')

plt.tight_layout()
plt.show()
```

```python
#(b) Apply the aforementioned intensity transformation to the saturation plane.


# Apply intensity transformation to saturation
def vibrance_boost(s_plane, a=0.8, sigma=70):
    s = s_plane.astype(np.float32)
    x = s.copy()
    boost = a * 128 * np.exp(-((x - 128) ** 2) / (2 * sigma ** 2))
    enhanced = np.minimum(x + boost, 255)
    return enhanced.astype(np.uint8)
```

```python

#(c) Adjust a to get a visually pleasing output. Report the value of a.

# Adjust `a` and report
#a_val = 0.8
a_val = 0.6
#a_val = 1
s_enhanced = vibrance_boost(s, a=a_val)
```

```python
#(d) Recombine the three planes.
# Step (d): Recombine and convert to RGB
hsv_enhanced = cv2.merge([h, s_enhanced, v])
img_enhanced_rgb = cv2.cvtColor(hsv_enhanced, cv2.COLOR_HSV2RGB)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#(a) Open the image in Fig. 5, split it into hue, saturation, and values and display these planes in grayscale

# Load and convert to HSV
image = cv2.imread("images/jeniffer.jpg")  
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
hue, sat, val = cv2.split(hsv)

# Show H, S, V planes
plt.figure(figsize=(12, 4))
for i, (channel, title) in enumerate(zip([hue, sat, val], ["Hue", "Saturation", "Value"])):
    plt.subplot(1, 3, i+1)
    plt.title(title)
    plt.imshow(channel, cmap='gray')
    plt.axis('off')
plt.suptitle("Hue, Saturation, Value Planes")
plt.show()
```

```python
# ✅ Save the transformed image to disk
cv2.imwrite("images/transformed/jeniffer_transformed_04_a.jpg", channel)
```

```python
#(b) Select the appropriate plane to threshold in extract the foreground mask. A mask is a binary image.

# Create a binary mask (based on value channel, can tweak threshold)
_,mask = cv2.threshold(val, 60, 255, cv2.THRESH_BINARY)

plt.subplot(1, 3, 2)
plt.title("Mask")
plt.imshow(mask, cmap='gray')
plt.axis('off')
```

```python
#(c) Now obtain the foreground only using cv.bitwise_and and compute the histogram.

# Extract foreground using bitwise_and
foreground = cv2.bitwise_and(val, val, mask=mask)

plt.subplot(1, 3, 2)
plt.title("foreground")
plt.imshow(foreground, cmap='gray')
plt.axis('off')
```

```python
#(d) Obtain the cumulative sum of the histogram using np.cumsum.

# Compute histogram and CDF of the foreground
hist = cv2.calcHist([foreground], [0], mask, [256], [0, 256])
cdf = np.cumsum(hist).astype(np.float32)
cdf_normalized = cdf * 255 / cdf[-1]  # Normalize to [0, 255]

print(cdf_normalized)
```

```python
#(e) Use the formulas in slides to histogram-equalize the foreground.

# Equalize foreground using calculated CDF
equalized_foreground = cdf_normalized[foreground].astype(np.uint8)

plt.subplot(1, 3, 2)
plt.title("foreground")
plt.imshow(equalized_foreground, cmap='gray')
plt.axis('off')
```

```python
#(f) Extract the background and add with the histogram equalized foreground.
#  Merge back with background
background = cv2.bitwise_and(val, val, mask=cv2.bitwise_not(mask))
equalized_val = cv2.add(background, equalized_foreground)

# Merge HSV and convert to BGR
hsv_equalized = cv2.merge([hue, sat, equalized_val])
image_equalized = cv2.cvtColor(hsv_equalized, cv2.COLOR_HSV2BGR)


```
