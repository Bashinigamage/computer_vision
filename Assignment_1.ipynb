```python
import cv2
import numpy as np
import matplotlib.pyplot as plt



# Load grayscale image
image = cv2.imread('images/emma.jpg', cv2.IMREAD_GRAYSCALE)

print(image)

```

```python
# Create empty output image
transformed = np.zeros_like(image, dtype=np.uint8)

# Apply transformation based on ranges
for i in range(image.shape[0]):
    for j in range(image.shape[1]):
        pixel = image[i, j]
        if pixel < 50:
            new_val = pixel
        elif pixel < 100:
            new_val = 2 * pixel - 50
        elif pixel < 150:
            new_val = 2.5 * pixel - 150
        else:
            new_val = pixel + 100
        # Clip to 0–255
        transformed[i, j] = np.clip(new_val, 0, 255)

# Show results
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title("Original Image")
plt.imshow(image, cmap='gray')
plt.subplot(1, 2, 2)
plt.title("Transformed Image")
plt.imshow(transformed, cmap='gray')
plt.show()
```

```python
# ✅ Save the transformed image to disk
cv2.imwrite("images/transformed/transformed_image.jpg", transformed)
```
